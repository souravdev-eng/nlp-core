# 🔍 Named Entity Recognition (NER) with NLTK

**Big idea:** NER finds the **names of real-world things** in text — people, places, organizations — and labels what kind of thing each one is.

> 🖍️ Think of reading a news article with a highlighter: yellow for **people**, green for **places**, blue for **companies**.

**Common entity types:**
- **Person** — Gustave Eiffel
- **Location / country** — India, Paris
- **Organization** — Google, United Nations
- **Date / time** — 24-09-1989, 4:30pm
- **Money / percent** — $1 million, 20%

**Real-life use cases:** pulling names and companies out of news, sorting support tickets, filling forms from documents (resumes, invoices), smarter search.

### Step 1 — Our sample sentence
A sentence about the Eiffel Tower. It has a **person** (Gustave Eiffel), a **landmark** (Eiffel Tower) and **years** (1887, 1889).

```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

### Step 2 — Tokenize and POS-tag
NER builds on the previous notebooks:
1. `word_tokenize` → sentence ➜ words (notebook 1)
2. `pos_tag` → label each word as noun, verb, … (notebook 5)

The NER model uses these tags as clues: proper nouns (`NNP`) are often names.

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
tag_elements=nltk.pos_tag(words)
```

### Step 3 — Download the NER model and word list
- `maxent_ne_chunker` → the pre-trained NER model (newer NLTK versions use `maxent_ne_chunker_tab`)
- `words` → a big English word list the model uses as a reference

### Step 4 — Find the entities (`ne_chunk`)
`nltk.ne_chunk(tagged_words)` groups words into entities and labels them. `.draw()` opens the result as a tree picture in a **separate pop-up window** (it won't appear inside the notebook).

💡 To see it inside the notebook instead, use `print(nltk.ne_chunk(tag_elements))`.

**What it finds:**
- `Gustave Eiffel` ➜ **PERSON** ✅
- `Eiffel Tower` ➜ **ORGANIZATION** ❌ (it's a landmark — NLTK has no "landmark" label, so it guesses)
- `1887`, `1889` ➜ not labelled — NLTK's NER **doesn't detect dates**

**NLTK's labels:** `PERSON`, `ORGANIZATION`, `GPE` (country/city), `LOCATION`, `FACILITY`.

👉 For better results (including dates and money), real projects use **spaCy** or transformer models.

```python
nltk.ne_chunk(tag_elements).draw()
```

## 📝 Quick recap (cheat sheet)

| Step | Tool | What it does |
|---|---|---|
| 1 | `word_tokenize` | sentence ➜ words |
| 2 | `pos_tag` | words ➜ (word, tag) pairs |
| 3 | `ne_chunk` | tagged words ➜ named entities |

**Golden rule:** tokenize ➜ POS tag ➜ NER. Good for a quick start; use spaCy for serious work.
