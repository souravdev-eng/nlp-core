# 🏷️ Parts of Speech (POS) Tagging with NLTK

**Big idea:** Every word plays a **role** in a sentence — noun, verb, adjective, etc. **POS tagging** labels each word with its role (its **tag**).

> 🎭 Think of a play: the same actor can play different roles. `book` is a **noun** in *"read a book"* but a **verb** in *"book a ticket"*. The tagger looks at the context to decide.

**Example:** `"Taj Mahal is a beautiful Monument"`

| Word | Tag | Meaning |
|---|---|---|
| Taj, Mahal | `NNP` | proper noun (a name) |
| is | `VBZ` | verb (he/she/it form) |
| a | `DT` | determiner |
| beautiful | `JJ` | adjective |
| Monument | `NN` | noun |

**Real-life use cases:**
- Finding names, places, products (nouns) in text
- Giving the lemmatizer the correct `pos` (notebook 3!)
- Grammar checkers, chatbots, question answering

### 📖 The 5 tag families to remember
The first letters tell you the family:

- `NN…` = **noun** (desk, India)
- `VB…` = **verb** (take, took, taking)
- `JJ…` = **adjective** (big, beautiful)
- `RB…` = **adverb** (very, closely)
- `PRP…` = **pronoun** (I, my)

### Step 5 — Tag every sentence

For each sentence: split into words ➜ remove stopwords ➜ `nltk.pos_tag(words)` gives **(word, tag)** pairs.

**Notice in the output:**
- ✅ Right: `India` ➜ `NNP`, `captured` ➜ `VBD`
- ❌ Wrong: `looted` ➜ `JJ` (should be a verb), `father` ➜ `RB` (should be a noun)

**Why wrong?** The tagger guesses from the **surrounding words**. Removing stopwords breaks the grammar (*"came and looted us"* ➜ *"came looted us"*).

💡 Tag the **full sentence first**, then remove stopwords.

```python
## We will find the Pos Tag

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[word for word in words if word not in set(stopwords.words('english'))]
    #sentences[i]=' '.join(words)# converting all the list of words into sentences
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)
```

### Step 6 — A simple example

### Step 7 — Tag it
- `Taj`, `Mahal` ➜ `NNP` (proper noun — a name)
- `is` ➜ `VBZ` (verb, he/she/it form)
- `a` ➜ `DT` (determiner)
- `beautiful` ➜ `JJ` (adjective)
- `Monument` ➜ `NN` (noun)

All correct ✅ — because the full sentence (with stopwords `is` and `a`) was kept.

```python

print(nltk.pos_tag("Taj Mahal is a beautiful Monument".split()))
```

## 📝 Quick recap (cheat sheet)

| Item | What it is |
|---|---|
| `nltk.pos_tag(words)` | list of words ➜ list of (word, tag) pairs |
| Tag families | `NN*` noun · `VB*` verb · `JJ*` adjective · `RB*` adverb · `PRP*` pronoun |
| Model | `averaged_perceptron_tagger_eng` (download once) |

**Common mistakes:**
- Removing stopwords **before** tagging (breaks context ➜ wrong tags)
- Tagging single words without a sentence (no context to guess from)

**Golden rule:** tokenize ➜ **POS tag** ➜ then remove stopwords / lemmatize (with the right `pos`).

```python
import nltk
from nltk.corpus import wordnet, stopwords
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

wordnet_pos = {
    'J': wordnet.ADJ,# Adjective
    'V': wordnet.VERB, # Verb
    'N': wordnet.NOUN, # Noun
    'R': wordnet.ADV # Adverb
}

sentences = nltk.sent_tokenize(paragraph)

for i, sentence in enumerate(sentences):
    # Tokenize the sentence into words and remove stop words
    words = nltk.word_tokenize(sentence)

    # Remove stop words from the list of words
    words = [word for word in words if word not in stop_words]

    # Get the part-of-speech tags for the words
    pos_tags = nltk.pos_tag(words)

    # Lemmatize the words based on their part-of-speech tags
    lemmatized_words = []

    # Lemmatize each word based on its POS tag
    for word, tag in pos_tags:
        # Map the POS tag to the corresponding WordNet POS tag
        pos = wordnet_pos.get(tag[0], wordnet.NOUN)  # Default to noun if not found

        # Lemmatize the word using the lemmatizer and the mapped POS tag
        lemmatized_word = lemmatizer.lemmatize(word, pos=pos)

        # Append the lemmatized word to the list of lemmatized words
        lemmatized_words.append(lemmatized_word)
    
    sentences[i] = ' '.join(lemmatized_words)
```
