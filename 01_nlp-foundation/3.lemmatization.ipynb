# Lemmatization with NLTK

**Big idea:** Like stemming, **lemmatization** reduces words to a root form. But instead of blindly chopping endings, it **looks the word up in a dictionary** and returns a **real word** called the **lemma**.

> 📖 Stemming is a pair of scissors ✂️. Lemmatization is a dictionary 📖 — slower, but it knows `went`/`goes`/`gone` all mean **go**.

| Word | Stemming | Lemmatization |
|---|---|---|
| `history` | `histori` ❌ | `history` ✅ |
| `eaten` | `eaten` ❌ | `eat` ✅ |
| `goes` | `goe` ❌ | `go` ✅ |

**Real-life use cases:** Q&A systems, chatbots, text summarization — anywhere the output must be real, readable words.

### Step 1 — Setup: `WordNetLemmatizer`
- **WordNet** = a large English dictionary database that groups words by meaning and knows their base forms.
- `WordNetLemmatizer` = NLTK's tool that looks words up in WordNet to find the lemma.
- First time only: run `nltk.download('wordnet')` so the dictionary is available.

```python
## Q&A,chatbots,text summarization
from nltk.stem import WordNetLemmatizer
```

### Step 2 — Create the lemmatizer
Create one lemmatizer object and reuse it for every word.

```python
lemmatizer=WordNetLemmatizer()
```

### Step 3 — Tell it the **part of speech** (`pos`)

The lemmatizer needs to know **what kind of word** it is, because the root depends on it:

| `pos` | Meaning | Example |
|---|---|---|
| `'n'` | noun (**default**) | `cars` ➜ `car` |
| `'v'` | verb | `going` ➜ `go` |
| `'a'` | adjective | `better` ➜ `good` |
| `'r'` | adverb | `fairly` |

Here `going` with `pos='v'` ➜ `go` ✅

⚠️ If you leave `pos` out, it assumes **noun** — and `going` would stay `going`.

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
lemmatizer.lemmatize("going",pos='v')
```

### Step 4 — Same word list as the stemming notebook
So we can compare the results side by side.

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

### Step 5 — Lemmatize every word (as verbs)

**Notice in the output:**
- `eaten` ➜ `eat` ✅ — stemming couldn't do this!
- `history` ➜ `history` ✅ — stays a real word (stemmer gave `histori`)
- `finalized` ➜ `finalize` ✅ — a proper dictionary word
- `finally` ➜ `finally` — it's an adverb, not a verb, so nothing changes

```python
for word in words:
    print(word+" ----> "+lemmatizer.lemmatize(word,pos='v'))
```

### Step 6 — The word that beat both stemmers: `goes`
`goes` ➜ `go` ✅ (Porter and Snowball both gave `goe`). The dictionary knows irregular verb forms.

```python
lemmatizer.lemmatize("goes",pos='v')
```

### Step 7 — Adverbs stay as they are
- `fairly` (as verb) ➜ `fairly`
- `sportingly` (default noun) ➜ `sportingly`

Lemmatization only returns **dictionary base forms** — it won't cut `fairly` to `fair` like Snowball did, because they're different words in the dictionary.

💡 Wrong or missing `pos` = no change. In real projects, a **POS tagger** (`nltk.pos_tag`) picks the `pos` automatically.

```python
lemmatizer.lemmatize("fairly",pos='v'),lemmatizer.lemmatize("sportingly")
```

---
## 📝 Quick recap (cheat sheet)

| | Stemming | Lemmatization |
|---|---|---|
| How | Chops endings with rules | Looks up a dictionary (WordNet) |
| Output | Stem (may not be real: `histori`) | Lemma (always a real word) |
| Irregular forms (`goes`, `eaten`) | ❌ Missed | ✅ Handled |
| Needs `pos`? | No | Yes (default = noun) |
| Speed | ⚡ Fast | 🐢 Slower |

**Golden rule:** need **speed** ➜ stemming. Need **correct, readable words** ➜ lemmatization (and pass the right `pos`).
