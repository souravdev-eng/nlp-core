# 📘 Tokenization with NLTK

**Big idea:** Computers can't "read" a paragraph the way we do. First we have to **chop the text into small pieces** (sentences, then words). This chopping is called **tokenization**, and each piece is a **token**.

> 🍕 Think of a pizza: you can't eat it whole, so you cut it into slices (sentences), then bites (words).

---

### Step 1 — Setup
- `import nltk` → load the **NLTK** toolbox (Natural Language Toolkit), a Python library full of ready-made language tools.
- `nltk.download('punkt_tab')` → download **Punkt**, NLTK's "rule book" for knowing where sentences end. Without it, the sentence splitter won't work.
- It only downloads once; after that it says *"already up-to-date"*.

```python
import nltk
nltk.download('punkt_tab')
print("Hello, World!")
```

### Step 2 — Our sample text (the "corpus")
**Corpus** = just a fancy word for *"the text we're working on"*. Here it's a short paragraph we'll chop up in the next cells.

`"""..."""` (triple quotes) lets a Python string span several lines.

```python
corpus = """Hello welcome, to saurav's world. 
This is a sample text for tokenization. 
We will use NLTK library for tokenization. 
NLTK is a powerful library for natural language processing in Python. 
It provides easy-to-use interfaces to over 50 corpora and lexical resources such as WordNet, along with a suite of text processing libraries for classification, tokenization, stemming, tagging, parsing, and semantic reasoning. 
Let's get started with tokenization using NLTK!
"""
```

### Step 3 — Split the paragraph into **sentences** (`sent_tokenize`)

**What:** Takes a big paragraph → gives back a **list of sentences**.

**Why not just split on `.`?** Because a full stop doesn't always end a sentence:
- `Dr. Smith arrived.` → a naive split would break after "Dr."
- `It costs $3.50.` → would break inside the number

`sent_tokenize` is smart: it knows about abbreviations, numbers, `!` and `?`.

**Real-life use cases:**
- Summarising an article sentence by sentence
- Checking each sentence's sentiment (happy / angry review)
- Feeding text to a translator one sentence at a time

> 🧠 Remember: **sent** = **sent**ence. Paragraph ➜ sentences.

```python
# Tokenization using NLTK 
# Sentence to paragraphs
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(corpus)
print("Sentences:")
sentences
```

### Step 4 — What did we get back?
`type(...)` tells us the data type. The answer is **`list`** → a Python list where each item is one sentence (a string).

```python
type(sentences)
```

### Step 5 — Print each sentence
A simple `for` loop: go through the list and print one sentence per line, so we can see where NLTK made the cuts.

```python
for sentence in sentences:
    print(sentence)
```

### Step 6 — Split each sentence into **words** (`word_tokenize`)

**What:** Takes a sentence → gives back a **list of words and punctuation marks**.

**Notice in the output:**
- Punctuation becomes its own token: `world .` → `'world'`, `'.'`
- `saurav's` → `'saurav'`, `"'s"` (the "'s" is kept together because it means *"belongs to"*)
- `easy-to-use` stays as **one** word ✅

**Why we need it:** almost every NLP task works on words — counting word frequency, removing common words like "the", finding keywords, building a search engine, or feeding a machine-learning model.

> 🧠 Remember: **word_tokenize** = the "smart" word splitter. Sentence ➜ words.

```python
# Paragraph to words
from nltk.tokenize import word_tokenize
for sentence in sentences:
    print(word_tokenize(sentence))
print("Words:")
```

### Step 7 — `wordpunct_tokenize`: the "strict" splitter

**What:** Splits text into words **and splits on every punctuation mark**, no exceptions.

**Compare with `word_tokenize`:**
| Text | `word_tokenize` | `wordpunct_tokenize` |
|---|---|---|
| `saurav's` | `saurav`, `'s` | `saurav`, `'`, `s` |
| `easy-to-use` | `easy-to-use` | `easy`, `-`, `to`, `-`, `use` |

**When to use it:** when you want a simple, predictable rule ("letters together, symbols apart") — e.g. quick cleanup of messy text, codes, or logs.

> 🧠 Remember: **wordpunct** = cuts at **every punct**uation.

```python
from nltk.tokenize import wordpunct_tokenize
words = wordpunct_tokenize(corpus)
words
```

### Step 8 — `TreebankWordTokenizer`: the "classic" splitter

**What:** Uses the rules from the **Penn Treebank** (a famous, hand-labelled English dataset used in research). `word_tokenize` is actually built on top of this.

**Notice in the output:** full stops stay stuck to words in the middle of the text → `'world.'`, `'tokenization.'`, `'Python.'`. Only the **very last** `.` would be split off.

**Why?** Treebank expects to receive **one sentence at a time** (so it only splits the final period). That's why the normal flow is:
`sent_tokenize` ➜ then `word_tokenize` / Treebank on each sentence.

**Contractions:** `Let's` → `Let`, `'s` (same as `word_tokenize`).

> 🧠 Remember: **Treebank** = the research-standard rules; give it sentences, not whole paragraphs.

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()

words = tokenizer.tokenize(corpus)
words
```

---
## 📝 Quick recap (cheat sheet)

| Tool | Input ➜ Output | Personality |
|---|---|---|
| `sent_tokenize` | paragraph ➜ sentences | Knows where sentences really end |
| `word_tokenize` | sentence ➜ words | Smart; keeps `easy-to-use`, splits `'s` |
| `wordpunct_tokenize` | text ➜ words | Strict; splits at every symbol |
| `TreebankWordTokenizer` | sentence ➜ words | Classic research rules; feed it one sentence |

**Golden rule:** first cut into **sentences**, then cut each sentence into **words**.
